```python
import matplotlib.pyplot as plt
import seaborn as sns
from pandas import read_csv
```

```python
df = read_csv('../data/processed/EDA2024.csv')
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['luot_nghe'], bins=50, kde=True)

plt.title('Phân phối lượt nghe trên Spotify Global Chart 2024')
plt.xlabel('Lượt nghe')
plt.ylabel('Số bài hát')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,4))
sns.boxplot(x=df['luot_nghe'])

plt.title('Boxplot lượt nghe – Phát hiện outliers')
plt.xlabel('Lượt nghe')

plt.tight_layout()
plt.show()
```

```python
top_songs = (
    df.groupby('ten_bai_hat')['luot_nghe']
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

plt.figure(figsize=(10,6))
top_songs.sort_values().plot(kind='barh')

plt.title('Top 10 bài hát có tổng lượt nghe cao nhất – Spotify 2024')
plt.xlabel('Tổng lượt nghe')
plt.ylabel('Bài hát')
plt.tight_layout()
plt.show()
```

```python
top_artists = (
    df['nghe_si']
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10,6))
sns.barplot(x=top_artists.values, y=top_artists.index)

plt.title('Top 10 nghệ sĩ xuất hiện nhiều nhất trên Spotify Global Chart 2024')
plt.xlabel('Số lần xuất hiện')
plt.ylabel('Nghệ sĩ')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(
    data=df,
    x='vi_tri_xep_hang',
    y='luot_nghe',
    alpha=0.4
)

plt.title('Mối quan hệ giữa vị trí xếp hạng và lượt nghe')
plt.xlabel('Vị trí xếp hạng (càng nhỏ càng cao)')
plt.ylabel('Lượt nghe')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['so_tuan_tren_bang_xep_hang'], bins=20, kde=True)

plt.title('Phân phối số tuần trụ hạng của các bài hát')
plt.xlabel('Số tuần trên bảng xếp hạng')
plt.ylabel('Số bài hát')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,6))
df['bai_hat_moi'].value_counts().plot(
    kind='pie',
    autopct='%1.1f%%',
    labels=['Không phải bài mới', 'Bài mới'],
    startangle=90
)

plt.title('Tỷ lệ bài hát mới trên Spotify Global Chart 2024')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df['thay_doi_thu_hang'], bins=30)

plt.title('Phân phối thay đổi thứ hạng giữa các tuần')
plt.xlabel('Thay đổi thứ hạng')
plt.ylabel('Số bài hát')
plt.tight_layout()
plt.show()
```
